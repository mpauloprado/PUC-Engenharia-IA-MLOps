# EXERCÍCIO 2
Considere o seguinte conjunto de dados:

- data = {'Tamanho': [150, 120, 180, 130, 160],
        'Quartos': [3, 2, 4, 3, 4],
        'Preço': [200000, 180000, 250000, 190000, 220000],
        'Localização': ['Norte', 'Leste', 'Oeste', 'Sul', 'Centro']}

Aplique a técnica de seleção de features para selecionar as features mais relevantes para o negócio.


```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_regression

dados = {'Tamanho': [150, 120, 180, 130, 160],
        'Quartos': [3, 2, 4, 3, 4],
        'Preço': [200000, 180000, 250000, 190000, 220000],
        'Localização': ['Norte', 'Leste', 'Oeste', 'Sul', 'Centro']}
df_dados = pd.DataFrame(dados)
df_dados_codificado = pd.get_dummies(df_dados, columns = ['Localização'], dtype=int, prefix='loc')
X = df_dados_codificado.drop('Preço', axis=1)
target = df_dados_codificado['Preço']

selector = SelectKBest(score_func=f_regression, k=2)
features = selector.fit_transform(X, target)

print(f'Features selecionadas: {X.columns[selector.get_support()]}')
```

# EXERCÍCIO 3

Considere o seguinte conjunto de dados:
- data = {'Matemática': [85, 92, 78, 96, 88],
        'Português': [7, 5, 8, 9, 6]}

É conhecido que as notas de Matemática variam de 0 a 100, enquanto as notas de Português variam de 0 a 10. Aplique a normalização Min-Max e a padronização Z-Score para as notas em ambas as disciplinas, de modo que elas estejam na mesma escala e distribuição.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

data = {'Matemática': [85, 92, 78, 96, 88],
        'Português': [7, 5, 8, 9, 6]}
df_dados = pd.DataFrame(data)

min_max_scaler = MinMaxScaler()
df_dados['Nota_Portugues_Normalizada'] = min_max_scaler.fit_transform(df_dados[['Português']]) 
df_dados['Nota_Matematica_Normalizada'] = min_max_scaler.fit_transform(df_dados[['Matemática']])
standard_scaler = StandardScaler()
df_dados['Nota_Portugues_Padronizada'] = standard_scaler.fit_transform(df_dados[['Português']]) 
df_dados['Nota_Matematica_Padronizada'] = standard_scaler.fit_transform(df_dados[['Matemática']])

df_dados
```

# EXERCÍCIO 6

Considere o conjunto de dados transacoes_bancarias.csv, disponibilizado no Canvas:
 * Aplique uma técnica de Feature Engineering para descobrir correlações entre Sexo e tipo de produto comprado.
* Aplique uma técnica de Feature Engineering para identificar qual dia da semana (domingo à sábado) ocorre mais compras.
* Aplique uma técnica de Feature Engineering para dimensionar os valores dos produtos por estado numa escala de 0 a 1.

```python

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

pd.set_option('display.max_rows', None)

df_dados = pd.read_csv("bases_dados/transacoes_bancarias.csv")

# Aplique uma técnica de Feature Engineering para descobrir correlações entre Sexo e tipo de produto comprado.
df_dados_agrupado = df_dados.groupby(['Sexo', 'Produto'])['Valor'].agg(quantidade='count').reset_index().pivot(index='Produto', columns='Sexo', values='quantidade').reset_index()
df_dados_agrupado['Perc_Sexo_Fem_sobre_Mas'] = (df_dados_agrupado['Feminino'] / df_dados_agrupado['Masculino'])
df_dados_agrupado

# Aplique uma técnica de Feature Engineering para identificar qual dia da semana (domingo à sábado) ocorre mais compras.
df_dados['Data'] = df_dados['Data'].astype('datetime64[ns]')
df_dados['dia_semana'] = df_dados['Data'].dt.day_name()
df_dados.groupby(['dia_semana'])['Valor'].agg(quantidade='count').reset_index()

# Aplique uma técnica de Feature Engineering para dimensionar os valores dos produtos por estado numa escala de 0 a 1.
min_max_scaler = MinMaxScaler()
df_agrupa_produto_estado = df_dados.groupby(['Estado', 'Produto'])['Valor'].agg(soma='sum').reset_index()
df_agrupa_produto_estado['soma_normalizada'] = min_max_scaler.fit_transform(df_agrupa_produto_estado[['soma']])
df_agrupa_produto_estado.sort_values(by=['Estado', 'soma_normalizada'])
```
