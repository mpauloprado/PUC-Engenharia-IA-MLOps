```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.feature_selection import SelectKBest, f_classif

# Carregar dataset
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target



# Selecionar as 5 melhores features
selector = SelectKBest(score_func=f_classif, k = 5)
X_selected = selector.fit_transform(X, y)



selector_features = X.columns[selector.get_support()]
print(f"Features selecionadas: {selector_features}")
```

```python
import pandas as pd
 
df = pd.DataFrame({

    "preco": [10, 20, 15],

    "quantidade": [2, 5, 3],

    "data": pd.to_datetime(["2024-01-01", "2024-03-15", "2024-07-20"])

})
 
# Criar feature derivada

df["valor_total"] = df["preco"] * df["quantidade"]
 
# Extrair informações de tempo

df["mes"] = df["data"].dt.month

df["dia_semana"] = df["data"].dt.day_name()
 
print(df)
 
```

```python
import numpy as np
import pandas as pd
 
df = pd.DataFrame({"renda": [1000, 5000, 20000, 100000]})
 
df["renda_log"] = np.log1p(df["renda"])
df["renda_sqrt"] = np.sqrt(df["renda"])
 
print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
 
df = pd.DataFrame({"cidade": ["SP", "RJ", "BH", "SP", "BH"]})
 
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(df[["cidade"]])
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out())
 
df = pd.concat([df, encoded_df], axis=1)
print(df)
```
